```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from google.colab import files
```

```python
uploaded = files.upload()
```

```python
filename = next(iter(uploaded))
image = Image.open(filename).convert('RGB')
image_array = np.array(image)
```

```python
image_array.shape
```

```python
plt.imshow(image_array)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
print("Image Shape:", image_array.shape)
print("Data Type:", image_array.dtype)
```

```python
print("Minimum Pixel Value:", image_array.min())
print("Maximum Pixel Value:", image_array.max())
```

```python
print("Image Height:", image_array.shape[0])
print("Image Width:", image_array.shape[1])
print("Number of Channels:", image_array.shape[2])
```

```python
# Inspect first pixel
print("First Pixel RGB:", image_array[0, 0])
```

Image Intensity}

```python
print("Image Shape:", image_array.shape)
print("Pixel at (0,0):", image_array[0, 0])
print("Pixel at (3,3):", image_array[3, 3])
```

```python
if image_array.shape[0] > 100 and image_array.shape[1] > 100:
    pixel = image_array[100, 100]

    print("Red intensity:", pixel[0])
    print("Green intensity:", pixel[1])
    print("Blue intensity:", pixel[2])
```

GrayScale Image

```python
img = np.array([
    [0, 50, 100, 150],
    [25, 75, 125, 175],
    [50, 100, 150, 200],
    [75, 125, 200, 255]
], dtype=np.uint16)
```

```python
plt.imshow(img, cmap='gray', vmin=0, vmax=255)
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

```python
print("Image Shape:", img.shape)
print("Pixel at (0,0):", img[0, 0])
print("Pixel at (3,3):", img[3, 3])
```
